# Explore results

Loads the scored dataset and reproduces the headline chart interactively. Pipeline scripts in `src/` are the source of truth.

```
import pandas as pd, plotly.express as px
df = pd.read_parquet('../data/clean/analysis_scored.parquet')
df[['instnm','firstgen_comp_orig_yr6_rt','predicted','residual_pct_pts','group']].head()
```

```
px.scatter(df, x='predicted', y='firstgen_comp_orig_yr6_rt', color='group', hover_name='instnm', opacity=.6)
```
